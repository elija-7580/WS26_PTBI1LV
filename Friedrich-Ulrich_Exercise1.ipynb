## Problem 6
Write code to count the digits of a given number n. The code should print the result to the console. For example, for n = 75869 the output should be 5.


```python



def count_digits(n):
    count = 0
    while n != 0:
        n = n // 10
        count = count + 1
    return count

n = 75869
print(count_digits(n))

print("\n")

n_str = str(n)
print(len(n_str))
```

## Problem 7

Write code to print all prime numbers between any given pair of numbers n and m to the console. Use as example input n = 25 and m = 50.
NOTE:A prime number is a natural number greater than 1 that can only be divided by itself and 1 without remainders. Examples are: 2, 3, 5, 7, 11.

n
m
x

    if x/EN except x =/ Elements of natural numbers

end while

```python
#prime numbers are natural numbers greater than 1 that have no positive divisors other than 1 and themselves. 

start=0 #inclusive
stop=50 #exclusive
step=1

list(range(start, stop, step))

for i in range(start, stop, step):
    if i > 1:  # Check if the number is greater than 1
        is_prime = True
        for j in range(2, int(i**0.5) + 1):  # Check for factors from 2 to sqrt(i)
            if i % j == 0:
                is_prime = False
                continue  # Not a prime number, continue to the next iteration
        if is_prime:
            print(i, "is a prime number")
```
